## 🧰 Ferramentas permitidas — e o que não pode usar

**Permitido neste relatório:**

- `for` sobre **listas**, **tuplas** e **strings**, inclusive com desempacotamento (`for a, b, c in lista_de_tuplas`);
- `range(stop)`, `range(start, stop)` e `range(start, stop, step)`, inclusive com passo **negativo**;
- `enumerate(colecao)` e `enumerate(colecao, start=n)`;
- `zip(a, b, ...)` e `enumerate(zip(...), start=n)`;
- *list comprehensions* **apenas nas Questões 8, 9 e 10**, onde o enunciado pede;
- `break`, `continue` e a cláusula `else` do `for`;
- listas e tuplas: indexação, fatiamento, `len()`, `append()`;
- métodos de texto `isdigit()`, `isalpha()` e `upper()`;
- `if`, `elif`, `else`, `try`, `except`, `raise`;
- `sum()` **somente nas Questões 8 e 9**, para totalizar uma lista gerada por comprehension;
- `print()`, `int()`, `float()`, `round()` e f-strings.

**Proibido neste relatório:**

- `while`;
- dicionários e conjuntos;
- funções próprias (`def` / `return`) — esse tópico será visto mais adiante;
- *list comprehensions* nas Questões 1 a 7 (use o `for` “por extenso” para ver o acumulador);
- `sum()`, `max()`, `min()` e `sorted()` fora do que foi liberado acima;
- NumPy, pandas ou qualquer biblioteca externa.

> 🚫 Se a solução usar uma estrutura proibida, a questão **não pontua**, mesmo que o resultado apareça. O objetivo é treinar o `for` e suas companheiras (`range`, `enumerate`, `zip`).

> 💡 **Regra de ouro:** o `for` repete **uma vez para cada item** de uma coleção finita — não há contador para esquecer nem loop infinito. Evite `for i in range(len(lista))` quando `enumerate(lista)` resolve.
